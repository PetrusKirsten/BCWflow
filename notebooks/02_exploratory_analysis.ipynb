# 02 — Exploratory Analysis v0

This notebook turns the processed queue-time dataset into the first operational EDA layer. 
At this stage, the goal is to validate analytical structure and generate reusable figures, not to claim stable operational conclusions from a tiny sample.


```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT / 'src'))

import pandas as pd
from parkflow.data.data_quality import load_best_available_dataset, build_coverage_summary
from parkflow.analysis.eda import (
    prepare_queue_dataset,
    build_attraction_summary,
    build_hourly_summary,
    build_heatmap_matrix,
    build_weather_summary,
    build_initial_insights,
)
from parkflow.visualization.plots import (
    plot_p90_wait_by_ride,
    plot_hourly_heatmap,
    plot_hourly_wait_profile,
    plot_weather_wait_comparison,
)
```

## 1. Load the best available processed dataset


```python
df_raw, path = load_best_available_dataset()
print(f'Loaded: {path}')
print(f'Rows: {len(df_raw):,}')
df = prepare_queue_dataset(df_raw)
df.head()
```

## 2. Coverage recap


```python
coverage = build_coverage_summary(df)
coverage
```

## 3. Conservative initial insights


```python
for item in build_initial_insights(df):
    print('-', item)
```

## 4. Attraction-level summary


```python
attraction_summary = build_attraction_summary(df, only_open=True)
attraction_summary.head(20)
```

```python
plot_p90_wait_by_ride(df, top_n=20)
```

## 5. Hourly patterns


```python
hourly_summary = build_hourly_summary(df, only_open=True)
hourly_summary
```

```python
plot_hourly_wait_profile(df)
```

## 6. Operational heatmap


```python
plot_hourly_heatmap(df, metric='mean_wait')
```

```python
heatmap_matrix = build_heatmap_matrix(df, metric='mean_wait', only_open=True)
heatmap_matrix.head()
```

## 7. Weather exploratory analysis


```python
weather_summary = build_weather_summary(df, only_open=True)
weather_summary
```

```python
plot_weather_wait_comparison(df)
```
